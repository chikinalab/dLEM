```python
from dlem.api import (jax_forward_generate, 
                        compute_contact_heatmap,
                        plot_map,
                        flip_diag_row)
from dlem.core import (normalize_expected_observed)
import jax 
```

```python
key = jax.random.PRNGKey(42)
n = 1000
resolution = 100
slowdown = .0025
rows = 200
```

```python
p_left = jax.random.uniform(key, shape=(n,))
p_right = jax.random.uniform(key, shape=(n,))
```

```python
out_data = jax_forward_generate(p_left, p_right, slowdown, rows)
```

```python
out_data.shape
```

```python
log = True
norm = normalize_expected_observed(out_data, log=log)
```

```python
import jax.numpy as jnp

padded_arr = jnp.pad(norm, 
                    pad_width=((0, n-rows)), 
                    mode='constant', 
                    constant_values=0)
prepared = flip_diag_row(padded_arr)
```

```python
chrom = "chr1"
start = 1
end = n*resolution
fig = plot_map(prepared, p_left, p_right, f'{chrom}:{start}-{end}', resolution)
```

```python
plot_width = 700
fig.update_layout(
    width=plot_width,
    height=plot_width*1.61803,
)
file_path = f"synthetic_data_plotly.html"
fig.write_html(file_path)
```
